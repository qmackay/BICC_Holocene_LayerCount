```python
#load GICC05

import pandas as pd
import numpy as np

file_path = '/Users/quinnmackay/Library/Mobile Documents/com~apple~CloudDocs/Programming/GitHub/BICC/Data Storage/Chronologies/gicc05_time_scale_excel.xlsx'
gicc05 = pd.read_excel(file_path, skiprows=40, usecols=[0,3,4])
```

```python
ng = gicc05.iloc[:,[0,1]].dropna()
grip = gicc05.iloc[:,[0,2]].dropna()
```

```python
grip_sections = [1529.489, 1532.799, 1536.493, 1538.572, 1539.06, 1540.919, 1541.411, 1542.558, 1544.894, 1549.833, 1574.659, 1575.828, 1585.05, 1598.901, 1609.428, 1615.623]

ng_sections = [1410.64, 1416.85, 1419.073, 1421.17, 1424.22, 1426.16, 1428.52, 1458.673, 1462.87, 1466.11, 1468.59, 1470.65, 1479.717, 1485.088]
```

```python
num_layers_total = []

for i in range(len(grip_sections)):
    if i == 0:
        num_layers_total.append(0)
        continue

    filter_grip = grip[(grip.iloc[:,1] >= grip_sections[i-1]) & (grip.iloc[:,1] < grip_sections[i])]
    num_layers = len(filter_grip)
    num_layers_total.append(num_layers)

pd.DataFrame((grip_sections, num_layers_total)).T.to_csv('/Users/quinnmackay/Desktop/grip_ng_count_out/grip_layer_counts.csv', index=False, header=False)
```
